```python
import datetime as dt
```

```python
today = dt.datetime.today()
print(today)
today = dt.date.today()
print(today)
```

```python
day = today.day
print(day)
```

```python
month = today.month
print(month)
```

```python
year = today.year
print(year)
```

```python
day_as_string = today.strftime("%A")
print(day_as_string)
```

```python
day_as_number = f"{today: %a}"
print(day_as_number)
```

```python
day_of_week = f"{today:%w}"
print(day_of_week)
```

```python
day_of_month = f"{today:%d}"
print(day_of_month)
```

```python
day_of_month_as_string = f"{today:%b}"
print(day_of_month_as_string)
```

```python
day_of_month_as_string = f"{today:%B}"
print(day_of_month_as_string)
```
