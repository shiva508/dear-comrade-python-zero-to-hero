```python
import datetime as dt
```

```python
class Person:
    description = "I am BATMAN"
    def __init__(self, first_name, last_name, age):
        self.first_name = first_name
        self.last_name = last_name
        self.age = age
        self.created_at = dt.datetime.now()
        self.status = False
    def show_date_of_joined(self):
        return f"{self.first_name} {self.last_name} joined {self.age} years old."
    def activate_profile(self, status):
        self.status = status
    @classmethod
    def update_description(cls, description):
        cls.description = description
    @staticmethod
    def current_date():
        return dt.datetime.now()
```

```python
user = Person('John', 'Doe', 22)
print(user)
print(user.first_name)
print(user.last_name)
print(user.age)
type(user)
user.first_name = 'Shiva'
print(user.first_name)
print(user.created_at)
print(user.show_date_of_joined())
print(user.status)
user.status = True
print(user.status)
print(user.description)
print(user.current_date())
print(Person.current_date())
user.update_description("Batman never give up")
print(user.description)
```

```python
class Batman(Person):
    pass
    def __init__(self, first_name, last_name, residency):
        super().__init__(first_name, last_name, 40)
        self.first_name = first_name
        self.last_name = last_name
        self.residency = residency
```

```python
person_batman = Batman(first_name='Bruce', last_name='Wayne', residency="Batcave")
print(person_batman.first_name)
print(person_batman.last_name)
print(person_batman.residency)
print(person_batman.status)
```
