# String indexing and slicing

```python
my_name = "Shiva Dasari"
print(my_name)
```

```python
my_name[0]
```

```python
my_name[-1]
```

```python
my_name[1:len(my_name):2]
```

```python
my_name[1:]
```

```python
my_name[:5]
```

```python
my_name[::2]
```
