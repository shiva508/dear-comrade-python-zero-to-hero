# Calculus

## Linear functions

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def liner_function(x):
    return 2*x
def square_function(x):
    return x**2
def exponential_function(x):
    return 2**x
```

```python
input_range = np.arange(-5, 5, 0.25)
liner_output_values = liner_function(input_range)
print(liner_output_values)
```

```python
plt.plot(input_range, liner_output_values)
```

```python
square_output_values = square_function(input_range)
plt.plot(input_range, square_output_values)
```

```python
exponential_output_values = exponential_function(input_range)
plt.plot(input_range, exponential_output_values)
```
