```python
class Sport:
    sport_type = "Outdoors"
    def __init__(self, name, number_of_players):
        self.name = name
        self.number_of_players = number_of_players
    def __str__(self):
        return f'{self.name} {self.number_of_players} players'
    def play_game(self):
        print(f"{self.name} {self.number_of_players} play now")
    def get_sport_type(self):
        return self.sport_type
```

```python
sport_obj = Sport(name="Cricket", number_of_players=11)
print(sport_obj)
sport_obj.play_game()
sport_obj.get_sport_type()
```

```python
class Cricket(Sport):
    sport_type = "Outdoors"
    def __init__(self, name, number_of_players, caption):
        super().__init__(name, number_of_players)
        self.caption = caption
    def get_caption_name(self):
        return self.caption
```

```python
cricket_obj = Cricket(name="Cricket", number_of_players=12, caption="Lara")
print(cricket_obj)
```

```python
cricket_obj.get_sport_type()
```
