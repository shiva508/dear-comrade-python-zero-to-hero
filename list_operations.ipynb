# List operations
## Lists are mutable sequences, typically used to store collections of homogeneous items.
## List can hold different data types like integers, strings, floats, etc.

```python
roll_numbers = [101, 102, 103, 104, 105]
print(roll_numbers)
```

```python
student_info = ["Shiva", 34, 3.75, True]
print(student_info)
```

```python
len(student_info)
```

```python
name = ["Shiva", "Dasari", "John", "Doe"]
min_name_length = min(len(item) for item in name)
print(min_name_length)
max_name_length = max(len(item) for item in name)
print(max_name_length)
```

# Index based access

```python
my_friend_names = ["Alice", "Bob", "Charlie"]
print(my_friend_names[0])
print(my_friend_names[1])
print(my_friend_names[2])
```

# Combining lists

```python
combining_list = roll_numbers + my_friend_names
print(combining_list)
```

# Slicing lists

```python
print(roll_numbers[1:4])
print(roll_numbers[2:len(roll_numbers)])
```

# Change list elements

```python
roll_numbers[0] = 508
print(roll_numbers)
```

# Adding elements to a list at the end using append()

```python
roll_numbers.append(408)
print(roll_numbers)
```

# Pop elements from a list using pop()
## Returns the last element and removes it from the list

```python
roll_numbers.pop()
print(roll_numbers)
```

# Sorting a list using sort()
## sort() method sorts the list in ascending order by default
## sort() method modifies the original list
## sort() method does not return a new list

```python
string_list = ["banana", "apple", "cherry", "date"]
string_list.sort()
print(string_list)

# Number list sorting
num_list = [34, 12, 5, 67, 23]
num_list.sort()
print(num_list)
```
