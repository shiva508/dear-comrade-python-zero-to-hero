```python
def square_root(num_input):
    return num_input * num_input
```

```python
my_nums = [2,4,6,8,10]
```

```python
map(square_root,my_nums)
```

```python
list(map(square_root,my_nums))
```

```python
def name_slicer(string_input):
    if len(string_input)%2:
        return string_input[2:]
    else:
        return string_input[:2]
```

```python
name_input = ["Shiva", "Satish", "Ravi"]
```

```python
list(map(name_slicer,name_input))
```
