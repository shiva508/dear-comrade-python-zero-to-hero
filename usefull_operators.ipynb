```python
"""
Range function
"""
"""
Using single parameter
"""
for num in range(10):
    print(num)
```

```python
"""
using start and stop parameters
"""
for num in range(0,11):
    print(num)
```

```python
"""
using start and stop, step parameters
"""
for num in range(0,11, 2):
    print(num)
```

```python
"""
enumerate operator
"""
name = "shiva"
for num in enumerate(name):
    print(num)
```

```python
name = "shiva"
for index, value in enumerate(name):
    print(f"index={index}, value= {value}")
```

```python
"""
Zip function
"""
team_a = [0,1,2,3,4]
team_b = [5,6,7,8,9]
combine_teams = zip(team_a, team_b)
print(combine_teams)
for item in combine_teams:
    print(item)
```
