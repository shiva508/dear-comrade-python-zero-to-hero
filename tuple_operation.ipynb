# Tuple Information
## Tuples are used to store multiple items in a single variable.
## A tuple is a collection which is ordered and unchangeable.
## Tuples are written with round brackets.

```python
user_info_tuple = ("Shiva", 34, 3.75, True)
print(user_info_tuple)
print(type(user_info_tuple))
```

```python
user_info_list = ["Shiva", 34, 3.75, True]
print(user_info_list)
print(type(user_info_list))
```

## Accessing tuple elements using indexing

```python
print(user_info_tuple[0])
print(user_info_tuple[1])
print(user_info_tuple[2])
```

## count() method to count occurrences of an element in a tuple

```python
sample_tuple = (1, 2, 3, 2, 4, 2, 5)
count_of_2 = sample_tuple.count(2)
print(count_of_2)
```

## index() method to find the index of the first occurrence of an element in a tuple

```python
sample_tuple = ("apple", "banana", "cherry", "banana", "date")
index_of_banana = sample_tuple.index("banana")
print(index_of_banana)
```

## Tuples are immutable - trying to change an element will raise an error

```python
sample_tuple[0] = "orange"
```
