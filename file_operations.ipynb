# File Operations
## File operations involve reading from and writing to files stored on your computer or other storage devices.
## They are essential for data persistence and manipulation in programming.

```python
names_file = open("names.txt", "w")
print(names_file)
names_file.seek(0)
```

## When file is not present, it creates a new file

```python
names_file = open("nam.txt")
names_file.seek(0)
```

## Reding a file

```python
names_file = open("users.txt")
file_content = names_file.read()
print(file_content)
names_file.seek(0)
```

```python
names_file = open("users.txt")
print(names_file.readlines())
names_file.seek(0)
```

## Gracefully closing a file

```python
with open("users.txt") as user_names_file:
    file_content = user_names_file.read()
    print(file_content)
```

## Writing to a file

```python
with open("users.txt", mode="w") as user_names_file:
    user_names_file.write("Shiva Dasari\n")
```

```python
with open("users.txt", mode="r") as user_names_file:
    file_content = user_names_file.read()
    print(file_content)
```

## Appending to a file

```python
with open("users.txt", mode="a") as user_names_file:
    user_names_file.write("John Doe\n")
```

```python
with open("users.txt", mode="r") as user_names_file:
    file_content = user_names_file.read()
    print(file_content)
```

## Mode explanation
### 'r' - Read mode (default): Opens a file for reading. Raises an error if the file does not exist.
### 'w' - Write mode: Opens a file for writing. Creates a new file if
### 'a' - Append mode: Opens a file for appending. Creates a new file if it does not exist.
### 'r+' - Read and write mode: Opens a file for both reading and writing.
### 'w+' - Write and read mode: Opens a file for both writing and reading. Creates a new file if it does not exist.
### 'a+' - Append and read mode: Opens a file for both appending and reading
### 'x' - Exclusive creation mode: Creates a new file and opens it for writing. Raises an error if the file already exists.
### 't' - Text mode (default): Used with other modes to handle text files (e.g., 'rt', 'wt').
### 'b' - Binary mode: Used with other modes to handle binary files (e.g., 'rb', 'wb').
