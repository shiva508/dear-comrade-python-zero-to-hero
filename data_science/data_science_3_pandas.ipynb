### Adding new column

```python
data_frame_csv["min_mix_score"] = 8
```

```python
data_frame_csv
```

```python
min_mix_score = np.arange(1,1001)
```

```python
data_frame_csv["min_mix_score"] = min_mix_score
```

```python
data_frame_csv["min_mix_score"]
```

```python
min_mix_score = np.random.randint(80,89, size=1000)
data_frame_csv["min_mix_score"] = min_mix_score
```

```python
data_frame_csv["min_mix_score"]
```

### 2nd way to add new column using assign()

```python
score_one = np.random.randint(80,89, size=1000)
score_two = np.random.randint(80,89, size=1000)
```

```python
series_one = pd.Series(score_one, index=np.arange(0,1000))
series_two = pd.Series(score_two, index=np.arange(0,1000))
```

```python
data_frame_csv.assign(score_one=series_one, score_two=series_two)
```

### 3rd way to add new column using insert()

```python
data_frame_csv.insert(1, "score_three", series_one)
```

```python
data_frame_csv
```
