```python
class Customer:
    def __init__(self, name, membership_type):
        self.name = name
        self.membership_type = membership_type
        print("Customer created")
    def set_membership_type(self, new_membership_type):
        self.membership_type = new_membership_type
    def set_name(self, new_name):
        self.name = new_name
```

```python
customer = Customer("Shiva", "Membership Type")
print(customer.membership_type)
customer.set_membership_type("GOLD")
print(customer.membership_type)
```

```python
customers = [Customer("Shiva", "Membership Type"),Customer("Shiva1", "Membership Type")]
for customer in customers:
    print(customer.name)
```
