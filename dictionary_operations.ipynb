# Dictionary Information
## Dictionaries are used to store data values in key:value pairs.
## A dictionary is a collection which is ordered*, changeable and do not allow duplicates.
## Dictionaries are written with curly brackets, and have keys and values.
## Without indexes, dictionaries use keys to access values.

## Dictionary: Unordered and accessing values via keys
## List: Ordered and accessing values via index and slicing operation can be done

```python
student_info = {
    "firstName": "Shiva",
    "lastName": "Dasari",
    "age": 34,
    "isGraduated": True,
    "courses": ["B.Tech", "M.Tech", "PhD"],
    "address": {
        "street": "123 Main St",
        "city": "Hyderabad",
        "state": "Telangana",
        "zip": "500001"
    }
}
print(student_info)
```

## Accessing dictionary values using keys

```python
print(student_info["firstName"])
print(student_info["lastName"])
print(student_info["age"])
print(student_info["isGraduated"])
print(student_info["courses"])
print(student_info["address"])
print(student_info["address"]["city"])
print(student_info["address"]["state"].upper())
```

## Modifying dictionary values

```python
modified_address = {
    "street": "456 Park Ave",
    "city": "Mumbai",
    "state": "Maharashtra",
    "zip": "400001"
}
student_info["address"] = modified_address
print(student_info)
```

## Getting all keys and values from a dictionary

```python
student_keys = student_info.keys()
print(student_keys)
student_values = student_info.values()
print(student_values)
```

## calling items() method to get key-value pairs

```python
student_info.items()
```
