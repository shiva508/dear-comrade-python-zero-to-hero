# Set Operations
## Sets are used to store multiple items in a single variable.
## A set is a collection which is unordered, unchangeable*, and unindexed.
## Sets are written with curly brackets.


```python
student_roll_numbers = set()
```

```python
print(student_roll_numbers)
```

```python
student_roll_numbers.add(408)
student_roll_numbers.add(423)
student_roll_numbers.add(450)
```

```python
print(student_roll_numbers)
```

## Adding duplicate elements to a set

```python
student_roll_numbers.add(408)  # Duplicate, will not be added
```

```python
print(student_roll_numbers)
```

### list to set conversion to remove duplicates

```python
number_list = [101, 102, 103, 101, 104, 102]
unique_numbers_set = set(number_list)
print(unique_numbers_set)
```
