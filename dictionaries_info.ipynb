```python
user = {
  "first_name": "John",
  "last_name": "Doe",
  "age": 22,
}

for user_value in user.values():
    print(user_value)
```

```python
user = {
  "first_name": "John",
  "last_name": "Doe",
  "age": 22,
}
for user_key in user.keys():
    print(f"{user_key}: {user.get(user_key)}")
```

```python
user_shiva = {
    "first_name": "Shiva",
    "last_name": "Dasari",
    "age": 22,
}
user_bruce = {
    "first_name": "Bruce",
    "last_name": "Wayne",
    "age": 23,
}
user = [user_bruce, user_shiva]
print(user)
for user in user:
    print(user)
```

```python
"""
Copy dictinary
"""
user_shiva = {
    "first_name": "Shiva",
    "last_name": "Dasari",
    "age": 22,
}
user_bruce = user_shiva.copy()
user_bruce.update({"age": 23})
print(user_bruce)
```

```python
"""
Delete dictinary
"""
user_shiva = {
    "first_name": "Shiva",
    "last_name": "Dasari",
    "age": 22,
}
print(user_shiva)
del user_shiva["age"]
print(user_shiva)
```

```python
"""
Clear dictinary
"""
user_shiva = {
    "first_name": "Shiva",
    "last_name": "Dasari",
    "age": 22,
}
user_shiva.clear()
print(user_shiva)
```

```python
user_shiva = {
    "first_name": "Shiva",
    "last_name": "Dasari",
    "age": 22,
    "intrest": [{"name": "Cricket", "level": "Pro"}],
}
print(user_shiva)
user_shiva.pop("age")
print(user_shiva)
```

```python
user_shiva = {
    "first_name": "Shiva",
    "last_name": "Dasari",
    "age": 22,
    "intrest": [{"name": "Cricket", "level": "Pro"}],
}
user_bruce = user_shiva.fromkeys(user_shiva.keys())
print(user_bruce)
```
