```python
user = {
  "first_name": "John",
  "last_name": "Doe",
  "age": 22,
}
is_field_exist = False
field_name = "first_namee"
if field_name in user:
  is_field_exist = True
print(is_field_exist)
```

```python
user = {
  "first_name": "John",
  "last_name": "Doe",
  "age": 22,
}
first_name = user.get("first_name")
print(first_name)
```

```python
user = {
  "first_name": "John",
  "last_name": "Doe",
  "age": 22,
}
user.update({"age": 22})
print(user)
```
